```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl
mpl.rcParams['figure.dpi'] = 300
```

```python
ym = 1
xm = 3
X, Y = np.meshgrid(np.linspace(0, xm, 100),
                  np.linspace(0, ym, 100))
```

```python
Z = np.zeros_like(X)
Z += np.exp((-((X-2.5)/0.5)**2 - ((Y - 0.8)/(1/3))**2))
Z += np.exp((-((X-xm*0.55)/0.5)**2 - ((Y - ym/2)/0.5)**2))
Z += np.exp((-((X-0.5)/0.5)**2 - ((Y - 0.75)/0.5)**2))
Z += np.exp((-((X-0.2)/0.5)**2 - ((Y - 0.1)/(1/3))**2))
Z += np.exp((-((X-3)/0.5)**2 - ((Y - 0.1)/(1/3))**2))
```

```python
fig, ax = plt.subplots(figsize = (10, 3))
# plt.imshow(Z, cmap = "gray_r", clim = [0, 2])
ax.contour(X, Y, Z+0.15, levels = np.linspace(0, 1, 5), colors = 'k', linewidths = 0.75)
ax.contourf(X, Y, Z, levels = np.linspace(0.17, 1.7, 15), cmap = "gray_r")
ax.axis('off')
ax.patch.set_alpha(0)
ax.set_xlim([0.05, 2.95])
ax.set_ylim([0.05, 0.95])
fig.tight_layout()

fig.savefig("header.png", transparent = True)
```
